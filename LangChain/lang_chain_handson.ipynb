```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
from dotenv import load_dotenv

print('imported successfully...')
```

```python
# load environment file
load_dotenv()

# OPENAI_API_KEY = os.environ.get('OPENAI_API_KEY')

print('API Key has been set in the environment')
# print(OPENAI_API_KEY)
```

```python
# define the prompt template with placeholder
prompt_template = ChatPromptTemplate.from_messages([
    (
        'system',
        '''You are an expert e-commerce copywriter with 10 years of experience
        writing compelling product description from online shopping platforms.
        Your descriptions are clear, engaging, benifit-focused, and persuasive.
        Always write in a friendly yet professional tone.
        '''
    ),
    (
        'human',
        '''Write a product description for an e-commerce listing.

        Product Name: {product_name}
        Key Features: {features}
        Target Audience: {target_audience}

        Requirements:
        - Start with a catchy one-line hook
        - Highlight 3-4 key benefits (not just features - explain WHY each matters)
        - End with a short call-to-action
        - Keep the total lenght between 80-120 words
        - Use simple, everyday language
        '''
    )
])

print('prompt template created successfully....')
print()
print('Template Preview')
print('System Role: Expert e-commerce copywriter')
print('Input variable: {product_name}, {features}, {target_audience}')
print('Output Format: Hook --> Benefits --> Call-to-action')
```

```python
# Intialize the GPT Model via LangChain
llm = ChatOpenAI(
    model='gpt-3.5-turbo', # GPT model
    temperature = 0.7 # creativity level: 0 -> robatic, 1 -> very creative
)
```

```python
# initialize output parser
output_parser = StrOutputParser()
```

```python
# build the chain by connection 3 steps using pipe symbol |
chain = prompt_template | llm | output_parser
```

```python
# sample product
product_input = {
    'product_name' : 'ProSound X1 Wireless Bluetooth Earbuds',
    'features' : '30-hours total battery life, Active Noise Cancellation (ANC), IPX5 waterproof, 10-minutes fast charge 2 hours playback, Bluetooth 5.3',
    'target_audience' : 'college students, gym-goers, and daily commuters who want premium audio without the premium price'
}
```

```python
# call llm through chain
description = chain.invoke(product_input)

# llm output
print(description)
```

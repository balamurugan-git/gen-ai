# List
Ordered list, mutable and allows duplicate

```python
vechicles = ['car', 'bike']
print(vechicles)

vechicles.append('cycle')
print(vechicles)

print(vechicles[1]) # accessing the list

vechicles[0] = 'bicycle' # modifying the existing data

print(vechicles)

print(len(vechicles)) # displaying lenght of vechicles
```

# array
Cannot store different data type
Mutable,  all duplicate, Homogeneous Data

```python
import array

temp = array.array('f', [23.4, 13.5, 22.7, 45.0]) # f indiates float type
print(temp)

temp.append(11.9)
print(temp)

# array to list
temp_list = list(temp)
print(temp_list)

# to tuple
temp_tuple = tuple(temp)
print(temp_tuple)

# to Array again
temp_array = array.array('f', temp_tuple)
print(temp_array)
```

# stack
LIFO - Last in, First out

```python
browser_history = []
browser_history.append('google.com')
browser_history.append('github.com')
print(browser_history)

last_element = browser_history.pop()
print(last_element)
print(browser_history)
```

# queue
FIFO - Fist in, First Out

```python
# ticket processing, ordered processing, messaging system
# avoid outage

from collections import deque
ticket_queue = deque()
ticket_queue.append('Ram')
ticket_queue.append('Gokul')
ticket_queue.append('Kalai')

print(ticket_queue)

first_person = ticket_queue.popleft()
print(first_person)

# why pop cannot be used for LIFO (Stack)?
last_person = ticket_queue.pop()
print(last_person)
print(ticket_queue)
```

# named tuple


```python
from collections import namedtuple

finger_print = namedtuple('FingerPrint',['id','datetime'])

fp = finger_print('1021','2026-01-02T05:00:00.000')

print(fp)

print(fp)
print('user id',fp.id)
print('date time', fp.datetime)
```

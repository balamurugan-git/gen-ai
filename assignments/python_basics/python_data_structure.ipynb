## Q1 — List + Dictionary: Student Marks Tracker

You are building a simple marks tracker for a class of students.

**Tasks:**
1. Create a **list** of 5 student names of your choice
2. Create a **dictionary** that maps each student name to a list of 3 marks (out of 100) — one mark per subject: `Maths`, `Science`, `English`
3. Loop through the dictionary and print each student's **name** and their **average mark** (rounded to 1 decimal place)
4. Find and print the student with the **highest average**
5. Add a **new student** with 3 marks to the dictionary and print the updated count of students

**Expected output format:**
```
Arjun   → Average: 87.3
Priya   → Average: 91.0
...
Top student: Priya with 91.0
Total students after addition: 6
```

````python
students = ['Karunan', 'Xavier', 'Amit', 'Ameer', 'Kusheek']
student_mark = {
    'Karunan' : [89, 78, 64],
    'Xavier' : [79, 90, 80],
    'Amit' : [81, 60, 70],
    'Ameer' : [75, 95, 99],
    'Kusheek' : [66, 78, 80]
}

def calculate_average( student_mark : dict ) -> dict:
    student_avg_map ={}
    for student, mark in student_mark.items():
        average = round( (sum(mark) / len(mark)) , 1)
        student_avg_map[student] = average
        print(f'{student} -> Average: {average}')

    return student_avg_map

def find_top_student(result : dict) -> dict:    
    output = {
        'top_student' : max(result, key=result.get),
        'top_avg' : max(result.values())
    }

    print(f'Top student: {output['top_student']} with {output['top_avg']}')
    return output


result = calculate_average(student_mark)
output = find_top_student(result)

# add new student and mark
students.append('Kabeer')
student_mark['Kabeer'] = [89, 97, 95]

# Display updated results
# Total Students
print('\n────────────────── Updated Result ──────────────────\n')
print(f'Total students after addition: {len(students)}')
result = calculate_average(student_mark)
output = find_top_student(result)
````

## Q2 — Set Operations: Course Enrollment System

A training institute has three batches. Each batch has a set of enrolled students.

**Given data (copy this into your cell):**
```python
batch_morning   = {'Arjun', 'Priya', 'Ravi', 'Meena', 'Kumar'}
batch_evening   = {'Priya', 'Suresh', 'Meena', 'Divya', 'Arjun'}
batch_weekend   = {'Ravi', 'Divya', 'Anand', 'Priya'}
```

**Tasks:**
1. Find students enrolled in **all three batches** (intersection of all three)
2. Find students enrolled in **at least one batch** (union of all three)
3. Find students enrolled in **morning only** (not in evening or weekend)
4. Find students enrolled in **exactly two batches** (morning & evening, but not weekend)
5. Check if `'Divya'` is enrolled in the morning batch — print `True` or `False`
6. Print the **total number of unique students** across all batches

**Expected output format:**
```
In all 3 batches    : {'Priya'}
In at least 1 batch : {'Arjun', 'Priya', ...}
Morning batch only  : {'Kumar'}
...
```

````python
batch_morning   = {'Arjun', 'Priya', 'Ravi', 'Meena', 'Kumar'}
batch_evening   = {'Priya', 'Suresh', 'Meena', 'Divya', 'Arjun'}
batch_weekend   = {'Ravi', 'Divya', 'Anand', 'Priya'}

# Students in all batches
students_in_batches = batch_morning & batch_evening & batch_weekend
print(f'In all 3 batches : {students_in_batches}')

# Students in atleast one batches
students_in_atleast_one_batch = batch_morning | batch_evening | batch_weekend
print(f'In atleast 1 batch : {students_in_atleast_one_batch}')

# Students only in morning batch
students_only_in_morning_batch = batch_morning - batch_evening - batch_weekend
print(f'Morning batch only : {students_only_in_morning_batch}')

# Students in Morning and Evening batch but not in weekend batch
students_in_morning_evening_batch = (batch_morning & batch_evening) - batch_weekend
print(f'In Morning & Evening batch : {students_in_morning_evening_batch}')

# Verify Student: Divya present in morning batch
is_student_present_in_morning_batch = {'Divya'}.issubset(batch_morning)
print(f'Is Divya part of Morning batch : {is_student_present_in_morning_batch}')

# Total no. of unique students across all batches
total_student_across_batches = len(batch_morning | batch_evening | batch_weekend)
print(f'Total no. of students across all batches : {total_student_across_batches}')
````

## Q3 — Stack & Queue: Customer Support System

A customer support system handles two types of tickets:
- **VIP tickets** are handled with a **Stack** (most recent VIP issue handled first — LIFO)
- **Regular tickets** are handled with a **Queue** (first come, first served — FIFO)

**Tasks:**
1. Create an empty **list** called `vip_stack` and an empty **deque** called `regular_queue`
2. Add these VIP tickets to the stack (in this order): `'VIP-001'`, `'VIP-002'`, `'VIP-003'`
3. Add these regular tickets to the queue (in this order): `'REG-101'`, `'REG-102'`, `'REG-103'`
4. Process (remove and print) the **top VIP ticket** from the stack
5. Process (remove and print) the **first regular ticket** from the queue
6. Add two new tickets: `'VIP-004'` to the stack and `'REG-104'` to the queue
7. Print the **current state** of both the stack and queue after all operations

**Expected output format:**
```
Processing VIP ticket : VIP-003
Processing REG ticket : REG-101
VIP stack now   : ['VIP-001', 'VIP-002', 'VIP-004']
Regular queue now: deque(['REG-102', 'REG-103', 'REG-104'])
```

````python
from collections import deque

vip_stack = [] # Stack - LIFO
regular_queue = deque() # Queue _ FIFO

vip_stack.append('VIP-001')
vip_stack.append('VIP-002')
vip_stack.append('VIP-003')

regular_queue.append('REG-101')
regular_queue.append('REG-102')
regular_queue.append('REG-103')

removed_vip = vip_stack.pop()
print(f'Processing VIP ticket : {removed_vip}')

removed_requeue = regular_queue.popleft()
print(f'Processing REG ticket : {removed_requeue}')

vip_stack.append('VIP-004')
print(f'VIP stack now : {vip_stack}')

regular_queue.append('REG-104')
print(f'Regular queue now : {regular_queue}')
````

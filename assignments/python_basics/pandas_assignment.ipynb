**Q2. Filtering & Derived Columns**
Using the orders DataFrame from Q1:
- Add a `total_price` column (`quantity × unit_price`)
- Filter all orders where `total_price > 500`
- Find the customer with the highest single-order total

```python
# calculate total price
online_orders['total_price'] = online_orders['quantity'] * online_orders['unit_price']
print('--- after adding total price ---')
print(online_orders.head())
print(online_orders.describe())

# filter orders below 500 total price
price_below_500 = online_orders[ online_orders['total_price'] > 500]
print('\n--- Orders above total price of 500 ---')
print(price_below_500.head())

# find customer with highest single order
max_price_order = online_orders.iloc[online_orders['total_price'].idxmax()]
print('\n--- order with highest single order total ---')
print(max_price_order)
```

**Q3. Handling Missing Values**
Add `NaN` to at least 3 cells in your DataFrame (in different columns).
Then:
- Report which columns have missing values and how many
- Fill numeric missing values with the column median
- Fill text missing values with `'Unknown'`
- Verify no missing values remain

```python
# update few data as nan
np.random.seed(35)
random_index = np.random.choice(online_orders.index, 10)

print('--- selected data to nullify ---')
print(online_orders.loc[random_index, ['order_id', 'customer', 'product', 'quantity', 'unit_price', 'date', 'total_price']])

# nullify the data - excluding the order_id from nullifying
online_orders.loc[random_index, [ 'customer', 'product', 'quantity', 'unit_price', 'date', 'total_price']] = np.nan

print('\n--- after nullifying ---')
print(online_orders.loc[random_index, ['order_id', 'customer', 'product', 'quantity', 'unit_price', 'date', 'total_price']])
```

```python
# report on columns having missing value and how many
print('--- columns with missing value and count ---')
print(online_orders.isna().sum())
```

```python
# numeric columns with column median
# numeric columns - quantity, unit_price, total_price
# update the quantity, unit_price columns with column median - then reculate the total price
online_orders['quantity'].fillna(online_orders['quantity'].median(), inplace=True)
online_orders['unit_price'].fillna(online_orders['unit_price'].median(), inplace=True)
print('--- verifying the updated values ---')
print(online_orders.loc[random_index])
```

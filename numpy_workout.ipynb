```python
import numpy as np

arr1 = np.array([20,56,34,89,6,34]) # 1D array

print(arr1)

arr2 = np.array([
                [2,34,56,21,4],
                [2,34,56,21,4],
                [2,34,56,21,4]
                ]) # 2D array

print(arr2)

arr3 = np.zeros((3,4)) # empty array, 3 row and 4 columns with default value as 0
print(arr3)
```

```python
rng = np.arange(0, 20, 2) # build a array based on interval, in this - start with 0, till 20 (excluded) with interval of 2
print(rng)
```

```python
linsp = np.linspace(0, 1, 6) # build a array based on 2 number and split into n equal pieces
print(linsp)
```

```python
# array property
print('shape',arr2.shape)
print('dimension',arr2.ndim)
print('size',arr2.size)
print('data type',arr2.dtype)
print('memory',arr2.itemsize, 'bytes')
```

```python
# Indexing / slicing

arr4 = np.array([[12, 33, 55, 66],
                [13, 34, 53, 26],
                [11, 53, 25, 16]])

print(arr4)
# Index
'''
      0  1  2
   0 [1, 2, 3]
   1 [5, 6, 7]
   2 [4, 8, 9]
'''
print(arr4[0,1])
print(arr4[-1,-1])
```

```python
# Slicing
print('first 2 row', arr4[:2, :])

# slicing syntex
'''
     Row         Column
arr4[start:stop, start:stop]
if start value not specified then it takes from starting
if stop value not specified then it takes till the end
'''

# print('first 2 row', arr4[:2, :])
'''
print till 2nd row and all columns
'''

print('first 2 column', arr4[:, :2])
'''
print till all rows and till 2 column
'''
```

```python
# math operations
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])

print(a+b)
print(a-b)
```

```python
#scalar operation
print(a + 50)

#Universal functions
# square root 
print(np.sqrt(a))

# convert negative values to positive in array
c = [-2, 34, 54, -67]
print(np.abs(c))
```

```python
# Boolean masking
temps = np.array([20, 30, 40, 50, 60])

hot_city = temps > 35

print(hot_city)
print(temps[hot_city])

cold_city = temps < 25
print(cold_city)
print(temps[cold_city])
```
